```python
import re
import joblib
import nltk
# nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.stem.isri import ISRIStemmer
from sklearn.feature_extraction.text import TfidfVectorizer

# === Preprocessing steps

def remove_non_arabic(text):
    return re.sub(r'[^\u0620-\u064A\s]', ' ', text) if isinstance(text, str) else ''

def tokenization(text):
    return re.findall(r'\S+', text) if isinstance(text, str) else []

def remove_arabic_stopwords(tokens):
    arabic_stopwords = set(stopwords.words("arabic"))
    return [token for token in tokens if token not in arabic_stopwords]

def get_root(word, stemmer):
    return stemmer.stem(word) if isinstance(word, str) else word

def stemming_AR(tokens):
    stemmer = ISRIStemmer()
    return [get_root(token, stemmer) for token in tokens]

def rejoin_tokens(tokens):
    return ' '.join(token for token in tokens if isinstance(token, str) and token.strip())

def ASTD_pretraitement(text):
    if isinstance(text, str):
        text = remove_non_arabic(text)
        tokens = tokenization(text)
        tokens = remove_arabic_stopwords(tokens)
        tokens = stemming_AR(tokens)
        return rejoin_tokens(tokens)
    return ''
```

```python

# === Load vectorizer and model
vectorizer = joblib.load("tfidf_vectorizer_ASTD_Pre_Blc.pkl")  # adjust if needed
model = joblib.load("RandomForestClassifier_ASTD_Preprocessed_Balanced_best.pkl")

```
